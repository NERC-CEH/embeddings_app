```python
from airllm import AutoModel
```

```python
model = AutoModel.from_pretrained("meta-llama/Meta-Llama-3-8B", hf_token=HF_TOKEN)
```

```python
input_tokens = model.tokenizer(
    "Tell me a random fact.",
    return_tensors="pt",
    return_attention_mask=False,
    truncation=True,
    max_length=128,
    padding=False,
)
```

```python
generated_output = model.generate(
    input_tokens["input_ids"].cuda(),
    max_new_tokens=128,
    use_cache=True,
    return_dict_in_generate=True,
)
```
